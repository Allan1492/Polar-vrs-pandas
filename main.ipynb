```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

#descarga

```python


df = pd.read_csv("yellow_tripdata_2015-01.csv")
print(f"Filas: {df.shape[0]:,}, Columnas: {df.shape[1]}")
df.head()
```

#Describir las variables disponibles

```python
df.info()
df.describe()
```

#Identificar valores faltantes

```python
nulos = df.isnull().sum()
porcentaje = (nulos / len(df) * 100).round(2)
pd.DataFrame({'Nulos': nulos, 'Porcentaje': porcentaje}).sort_values('Porcentaje', ascending=False)
```

#Analizar la variable objetivo

```python
variable = 'total_amount'
print(df[variable].describe())

plt.figure(figsize=(10, 6))
plt.hist(df[(df[variable] > 0) & (df[variable] < 100)][variable], bins=50, edgecolor='black')
plt.title(f'Distribución de {variable}')
plt.xlabel('Monto ($)')
plt.ylabel('Frecuencia')
plt.show()
```
